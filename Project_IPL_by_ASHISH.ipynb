```python
# MISSING VALUES
matches['winner'].fillna("No result",inplace=True)

matches['city'].fillna("Unknown",inplace=True)

matches['player_of_match'].fillna("No Award",inplace=True)

matches['method'].fillna("Na",inplace=True)

matches['result_margin'].fillna("0",inplace=True)

matches['target_runs'].fillna("0",inplace=True)

matches['target_overs'].fillna("0",inplace=True)
```

```python
matches.isnull().sum()
```

```python
# Replaces old franches to new franches
matches.replace({
    'Delhi Daredevils':'Delhi Capitals',
    'Rising Pune Supergiants':'Rising Pune Supergiant'
}, inplace=True)

deliveries.replace({
    'Delhi Daredevils':'Delhi Capitals',
    'Rising Pune Supergiants':'Rising Pune Supergiant'
}, inplace=True)
```

```python
# CONVERT DATA COLUMN
matches['date'] = pd.to_datetime(matches['date'])
```

```python
# CREATE SEASON YEAR
matches['Season'] = matches['date'].dt.year
```

```python
plt.figure(figsize=(10,5))

sys.countplot(x='Season',data=matches)

plt.xticks(rotation=90)

plt.title("Matches Played Per Season")

plt.xlabel("Season")

plt.ylabel("Matches")

plt.show()

```

```python
season = matches['Season'].value_counts().sort_index()

plt.figure(figsize=(12,5))

plt.plot(season.index, season.values, marker='o')

plt.title("IPL Matches Trend")

plt.xlabel("Season")

plt.ylabel("Matches")

plt.show()
```

```python
plt.figure(figsize=(12,8))

matches['player_of_match'].value_counts().head(15).plot(kind='barh')

plt.title("Top Player of the Match Winners")

plt.show()
```

```python
matches['winner'].value_counts().head(8).plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(8,8)
)

plt.ylabel("")

plt.title("Top Winning Teams")

plt.show()
```

```python
# Fill missing values in deliveries dataset

deliveries['extras_type'] = deliveries['extras_type'].fillna('No Extra')

deliveries['player_dismissed'] = deliveries['player_dismissed'].fillna('Not Out')

deliveries['dismissal_kind'] = deliveries['dismissal_kind'].fillna('No Wicket')

deliveries['fielder'] = deliveries['fielder'].fillna('No Fielder')
```

```python
deliveries.isnull().sum()
```

```python
# DIAGRAM PRESENTATION OF DELIVERIES DATA
# TOP RUNS SCORER OF THE SEASON
top_batsmen = deliveries.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))

sys.barplot(
    x=top_batsmen.values,
    y=top_batsmen.index,
    palette='viridis'
)

plt.title("Top 10 Run Scorers in IPL")
plt.xlabel("Total Runs")
plt.ylabel("Batter")

plt.show()
```

```python
# TOP WICKET TAKER  OF SEASON
wickets = deliveries[deliveries['is_wicket']==1]

top_bowlers = wickets['bowler'].value_counts().head(10)

plt.figure(figsize=(12,6))

sys.barplot(
    x=top_bowlers.values,
    y=top_bowlers.index,
    palette='magma'
)

plt.title("Top 10 Wicket Takers")
plt.xlabel("Wickets")
plt.ylabel("Bowler")

plt.show()
```

```python
plt.figure(figsize=(6,6))
explode=[0,0.1]
deliveries['is_wicket'].replace(
    {0:'No Wicket',1:'Wicket'}
).value_counts().plot(
    kind='pie',explode=explode
    autopct='%1.1f%%'
)

plt.title("Wicket Distribution")

plt.ylabel("")

plt.show()
```

```python
# Matches Won
team_wins = matches['winner'].value_counts()

# Total Runs
team_runs = deliveries.groupby('batting_team')['total_runs'].sum()

comparison = pd.DataFrame({
    'Wins': team_wins,
    'Runs': team_runs
}).fillna(0)

comparison = comparison.sort_values('Wins', ascending=False)

comparison.plot(
    kind='bar',
    figsize=(15,6)
)

plt.title("Team Wins vs Total Runs")
plt.xlabel("Teams")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# BEST FEILDER 
top_fielders=(deliveries[deliveries['is_wicket']==1]
              .groupby('fielder')
              .size()
              .sort_values(ascending=False)
              .head(10)
             )


# PLOT
plt.figure(figsize=(12,6))

#BAR CHART
plt.plot(top_fielders.index,top_fielders.values,marker='o',label="Trand",color='blue')
plt.bar(top_fielders.index,top_fielders.values,color='pink')
plt.title('Top 10 Fielders by Wickets',fontsize=16)
plt.xlabel('fielder')
plt.ylabel('Number of Wicket')
plt.legend()
plt.grid(axis='y',linestyle='--',alpha=0.5)
plt.tight_layout()
plt.show()
```
